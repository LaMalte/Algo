```python
def moyenne_n(n):
    if n == 0:
        print("ERREUR : n doit etre superieur a 0")
        return None
    somme = 0
    for i in range(1, n + 1):
        somme = somme + i
    moyenne = somme / n
    return moyenne

n = 100000
print("La moyenne (ite) des ",n, "premiers nombres est :",moyenne_n(n))
```

```python
def somme_n_rec(n):
    if n == 0:
        return 0
    return somme_n_rec(n - 1) + n


def moyenne_n_rec(n):
    """a partir de la somme recursive"""
    if n == 0:
        print("ERREUR : n doit etre superieur a 0")
        return None
    return somme_n_rec(n) / n

n = 100
print("La moyenne (rec) des" ,n, "premiers nombres est :",moyenne_n_rec(n))
```

```python
def factn(n):
    if n == 0:
        return 1
    facto = 1
    for i in range (1, n + 1):
        facto = facto * i
    return facto

n = 6
print("Le factoriel (ite) de",n, "est :",factn(n))


def fact_rec(n):
    if n == 0:
        return 1
    return (fact_rec(n - 1) * n)

n = 6
print("Le factoriel (rec) de",n, "est :",fact_rec(n))
```

```python
def plus_base2_v1(x,y):
    """1er solution"""
    z = x + y 
    reste = z - (z // 2) * 2
    return reste

x = 5
y = 7
print("Le reste (v1) de ",x, " et " ,y, " est : ", plus_base2_v1(x,y))


def plus_base2_v2(x,y):
    z = x + y 
    if z % 2 == 0:
        reste = 0
    else:
        reste = 1
    return reste

x = 4
y = 7
print("Le reste (v2) de ",x, " et " ,y, " est : ", plus_base2_v2(x,y))


def plus_base2_v3(x,y):
    z = x + y 
    if z % 2 == 0:
        return 0
    else:
        return 1

x = 3
y = 7
print("Le reste (v3) de ",x, " et " ,y, " est : ", plus_base2_v3(x,y))
```

```python
def fib_rec(n):
    if n == 0:
        return 0 
    if n == 1:
        return 1
    return fib_rec(n -1) + fib_rec(n - 2)
    
n = 25
print("Fibonnaci (rec) de",n, "est :",fib_rec(n))


def fib_ite(n):
    if n == 0:
        return 0
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b
    
n = 25
print("Fibonnaci (ite) de",n, "est :",fib_ite(n))
```
